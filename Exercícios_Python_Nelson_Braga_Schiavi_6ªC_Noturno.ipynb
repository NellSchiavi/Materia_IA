## 20. Matriz Transposta
Dada uma matriz (lista de listas), criar sua matriz transposta.

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6]
]

linhas = len(matriz)        # quantidade de listas dentro da matriz
colunas = len(matriz[0])    # quantidade de elementos da primeira linha

print("Matriz original:")
for linha in matriz:
    print(linha)


# [i][j] vai para a posição [j][i].
# Por isso a nova matriz tem dimensões invertidas: colunas x linhas.
transposta = []
for j in range(colunas):
    nova_linha = []
    for i in range(linhas):
        nova_linha.append(matriz[i][j])
    transposta.append(nova_linha)

print("\nMatriz transposta:")
for linha in transposta:
    print(linha)
```

## 21. Busca Binária
Implementar a busca binária para encontrar um elemento em uma lista ordenada.

```python
def busca_binaria(lista, alvo):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        # // é usado ´para divisão inteira
        meio = (inicio + fim) // 2

        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1   # alvo está na metade da direita
        else:
            fim = meio - 1      # alvo está na metade da esquerda

    return -1   # o intervalo se esgotou sem encontrar

# A busca binária funciona em listas ordenadas que permite descartar metade dos elementos.
numeros = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
alvo = 23 # Valor alvo da busca.

posicao = busca_binaria(numeros, alvo)

if posicao != -1:
    print(f"O número {alvo} está no índice {posicao}")
else:
    print(f"O número {alvo} não está na lista")
```

## 22. Criptografia de Cifra de César
Função para criptografar uma mensagem usando a cifra de César.

```python
def criptografar_cesar(mensagem, deslocamento):
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():   # só letras são cifradas
            # ord() devolve o código numérico do caractere na tabela ASCII.
            # Uso 'A' ou 'a' como base para trabalhar com 0 a 25.
            base = ord('A') if caractere.isupper() else ord('a')
            posicao = ord(caractere) - base

            # O % 26 faz o alfabeto "dar a volta": depois do Z volta ao A.
            nova_posicao = (posicao + deslocamento) % 26

            # chr() faz o caminho inverso de ord(): número vira caractere.
            resultado += chr(base + nova_posicao)
        else:
            resultado += caractere   # espaços e pontuação passam intactos

    return resultado

mensagem = input("Digite a mensagem para criptografar: ")
chave = int(input("Digite a chave númerica de desbloqueio (ex: 3): "))

print(f"Mensagem original:   {mensagem}")
print(f"Mensagem cifrada:    {criptografar_cesar(mensagem, chave)}")
```

## 23. Descriptografar Cifra de César
Função para descriptografar uma mensagem cifrada com a cifra de César.

```python
def descriptografar_cesar(mensagem, deslocamento):
    # Reaproveitei a função anterior
    # cifrar com -3 é exatamente o inverso de cifrar com +3.
    return criptografar_cesar(mensagem, -deslocamento)

cifrada = input("Digite a mensagem cifrada: ")
chave = int(input("Digite o deslocamento usado: "))

print(f"Mensagem cifrada:      {cifrada}")
print(f"Mensagem decifrada:    {descriptografar_cesar(cifrada, chave)}")

# Teste de consistência: cifrar e decifrar deve devolver o texto original.
teste = "Python e legal"
print(f"\nTeste: {descriptografar_cesar(criptografar_cesar(teste, 5), 5)}")
```

## 24. Validação de CPF
Função para validar um CPF brasileiro.

```python
def validar_cpf(cpf):
    # Mantém apenas os dígitos ignorando pontos e traços da formatação.
    numeros = [int(c) for c in cpf if c.isdigit()]

    if len(numeros) != 11:
        return False

    # CPFs com todos os dígitos iguais (111.111.111-11) passariam no
    # cálculo matemático, mas serão invalidados pelo sistema.
    if numeros == numeros[0:1] * 11:
        return False

    # 1º dígito verificador os 9 primeiros números são multiplicados
    # por pesos de 10 até 2; o resto da divisão da soma por 11 define o dígito.
    soma = 0
    for i in range(9):
        soma += numeros[i] * (10 - i)
    digito1 = (soma * 10) % 11
    if digito1 == 10:
        digito1 = 0

    if digito1 != numeros[9]:
        return False

    # 2º dígito: mesma lógica, agora com 10 números e pesos de 11 até 2.
    soma = 0
    for i in range(10):
        soma += numeros[i] * (11 - i)
    digito2 = (soma * 10) % 11
    if digito2 == 10:
        digito2 = 0

    return digito2 == numeros[10]

cpf = input("Digite um CPF (com ou sem pontuação): ")

if validar_cpf(cpf):
    print("CPF VÁLIDO")
else:
    print("CPF INVÁLIDO")
```

## 27. Anagramas
Verificar se duas palavras são anagramas (formadas pelas mesmas letras).

```python
def sao_anagramas(palavra1, palavra2):
    # deixar padronizado com o lower que nem fiz antes removendo espaços e ignorando maiúsculas.
    p1 = palavra1.lower().replace(" ", "")
    p2 = palavra2.lower().replace(" ", "")

    # sorted() em uma string devolve a lista de caracteres em ordem.
    # Se as duas listas ordenadas forem iguais, as letras são as mesmas.
    return sorted(p1) == sorted(p2)

palavra1 = input("Digite a primeira palavra: ")
palavra2 = input("Digite a segunda palavra: ")

if sao_anagramas(palavra1, palavra2):
    print(f'"{palavra1}" e "{palavra2}" SÃO anagramas')
else:
    print(f'"{palavra1}" e "{palavra2}" NÃO são anagramas')
```

## 30. Jogo da Forca
Jogo simples da forca onde o usuário deve adivinhar uma palavra.

```python
import random

palavras = ["python", "notebook", "algoritmo", "variavel", "funcao"]
palavra = random.choice(palavras)   # escolhe um item aleatório da lista

# set() guarda valores únicos
letras_certas = set()
letras_erradas = set()
tentativas_restantes = 6

print("=== JOGO DA FORCA ===")
print(f"A palavra tem {len(palavra)} letras.\n")

while tentativas_restantes > 0:
    # Monta a exibição com a letra revelada se já foi acertada, senão "_".
    exibicao = ""
    for letra in palavra:
        if letra in letras_certas:
            exibicao += letra + " "
        else:
            exibicao += "_ "

    print(f"Palavra: {exibicao}")
    print(f"Erros: {' '.join(letras_erradas)} | Tentativas: {tentativas_restantes}")

    # set(palavra) são as letras distintas da palavra. Se todas já foram acertadas, o jogador venceu.
    if set(palavra) == letras_certas:
        print(f"\nParabéns! Você acertou: {palavra.upper()}")
        break

    palpite = input("Digite uma letra: ").lower()

    if palpite in letras_certas or palpite in letras_erradas:
        print("Você já tentou essa letra.\n")
    elif palpite in palavra:
        letras_certas.add(palpite)
        print("Acertou uma letra!\n")
    else:
        letras_erradas.add(palpite)
        tentativas_restantes -= 1
        print("Letra errada.\n")

if tentativas_restantes == 0:
    print(f"Suas tentativas acabaram. A palavra era: {palavra.upper()}")
```

## 31. Ordenação por Bolha
Implementar o algoritmo de ordenação por bolha (Bubble Sort).

```python
def bubble_sort(lista):
    # Cria uma cópia para não alterar a lista original.
    numeros = lista.copy()
    n = len(numeros)

    # A cada passada completa, o maior elemento restante vai até o fim
    for i in range(n):
        trocou = False

        # O "- i" evita recomparar a parte final, que já está ordenada.
        for j in range(n - 1 - i):
            if numeros[j] > numeros[j + 1]:
                # Aqui tem uma troca de posições usando atribuição múltipla.
                numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]
                trocou = True

        # Se uma passada inteira não fez trocas, e a lista já está ordenada
        # então pode encerrar antes de completar todas as voltas.
        if not trocou:
            break

    return numeros

original = [64, 34, 25, 12, 22, 11, 90]
print(f"Lista original: {original}")
print(f"Lista ordenada: {bubble_sort(original)}")
print(f"Original preservada: {original}")
```

## 32. Jogo da Velha
Jogo da velha para dois jogadores.

```python
def exibir_tabuleiro(tab):
    print()
    for i in range(3):
        print(f" {tab[i][0]} | {tab[i][1]} | {tab[i][2]} ")
        if i < 2:
            print("---+---+---")
    print()

def verificar_vitoria(tab, simbolo):
    for i in range(3):
        # Linha completa
        if tab[i][0] == tab[i][1] == tab[i][2] == simbolo:
            return True
        # Coluna completa
        if tab[0][i] == tab[1][i] == tab[2][i] == simbolo:
            return True

    # Diagonal principal e diagonal secundária
    if tab[0][0] == tab[1][1] == tab[2][2] == simbolo:
        return True
    if tab[0][2] == tab[1][1] == tab[2][0] == simbolo:
        return True

    return False

# Tabuleiro inicial numerado de 1 a 9 para facilitar a escolha do jogador.
tabuleiro = [["1", "2", "3"],
             ["4", "5", "6"],
             ["7", "8", "9"]]

jogador = "X"
jogadas = 0

print("=== JOGO DA VELHA ===")
print("Escolha a posição digitando o número correspondente.")

while True:
    exibir_tabuleiro(tabuleiro)
    posicao = input(f"Jogador {jogador}, escolha uma posição (1-9): ")

    if posicao not in "123456789" or posicao == "":
        print("Posição inválida.")
        continue   # volta ao início do laço sem trocar o jogador

    # Converto o número 1-9 em coordenadas de linha e coluna:
    # divisão inteira por 3 dá a linha, o resto dá a coluna.
    indice = int(posicao) - 1
    linha = indice // 3
    coluna = indice % 3

    if tabuleiro[linha][coluna] in ("X", "O"):
        print("Essa posição já foi usada. Tente outra.")
        continue

    tabuleiro[linha][coluna] = jogador
    jogadas += 1

    if verificar_vitoria(tabuleiro, jogador):
        exibir_tabuleiro(tabuleiro)
        print(f"O jogador {jogador} venceu!")
        break

    # 9 jogadas sem vitória significa que deu velha.
    if jogadas == 9:
        exibir_tabuleiro(tabuleiro)
        print("Deu velha! Ninguém venceu.")
        break

    # Alterna entre X e O usando um if de uma linha.
    jogador = "O" if jogador == "X" else "X"
```

## 36. Gerador de Senhas
Criar um gerador de senhas aleatórias com letras e números.

```python
import random
import string

def gerar_senha(tamanho):
    # string.ascii_letters = a-z + A-Z; string.digits = 0-9.
    # Utilizando essas constantes prontas não precisa digitar o alfabeto inteiro.
    caracteres = string.ascii_letters + string.digits

    # random.choice() sorteia um caractere
    senha = "".join(random.choice(caracteres) for _ in range(tamanho))
    return senha

tamanho = int(input("Quantos caracteres a senha deve ter? "))

print(f"\nSenha gerada: {gerar_senha(tamanho)}")

# está aqui gerando mais algumas para evidenciar que o resultado é sempre diferente e dando mais opções.
print("\nOutras opções:")
for _ in range(3):
    print(gerar_senha(tamanho))
```

## 37. Funções Recursivas
Implementar uma função recursiva para calcular o fatorial de um número.

```python
def fatorial_recursivo(n):
    if n <= 1:
        return 1

    # A função entende que a chamada menor devolve o resultado correto.
    return n * fatorial_recursivo(n - 1)

numero = int(input("Digite um número para calcular o fatorial: "))

print(f"{numero}! = {fatorial_recursivo(numero)}")
```

## 40. Lista de Compras
Programa para gerenciar uma lista de compras: adicionar, remover e listar itens.

```python
lista_compras = []

def mostrar_menu():
    print("\n=== LISTA DE COMPRAS ===")
    print("1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Listar itens")
    print("4 - Sair")

# O laço mantém o programa rodando até o usuário escolher sair.
while True:
    mostrar_menu()
    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        item = input("Nome do item: ")
        lista_compras.append(item)
        print(f'"{item}" adicionado.')

    elif opcao == "2":
        if not lista_compras:   # lista vazia é considerada False
            print("A lista já está vazia.")
        else:
            item = input("Qual item remover? ")
            if item in lista_compras:
                # .remove() apaga a primeira ocorrência do valor.
                lista_compras.remove(item)
                print(f'"{item}" removido.')
            else:
                print("Item não encontrado na lista.")

    elif opcao == "3":
        if not lista_compras:
            print("A lista está vazia.")
        else:
            print("\nItens da lista:")
            # enumerate() devolve o índice junto com o item;
            # start=1 faz a numeração começar em 1
            for i, item in enumerate(lista_compras, start=1):
                print(f"{i}. {item}")

    elif opcao == "4":
        print("Encerrando. Lista final:", lista_compras)
        break

    else:
        print("Opção inválida.")
```

## 42. Números Primos até N
Gerar todos os números primos até um número n fornecido.

```python
def eh_primo(numero):
    if numero < 2:
        return False
    for divisor in range(2, int(numero ** 0.5) + 1):
        if numero % divisor == 0:
            return False
    return True

n = int(input("Listar primos até qual número? "))

# Uso list comprehension com filtro para manter apenas os primos
primos = [numero for numero in range(2, n + 1) if eh_primo(numero)]

print(f"\nPrimos até {n}: {primos}")
print(f"Quantidade encontrada: {len(primos)}")
```

## 43. Conversor de Unidades

## 46. Calcular Média e Desvio Padrão
Calcular a média e o desvio padrão de uma lista de números.

```python
import statistics

notas = [7.5, 8.0, 6.5, 9.0, 7.0, 8.5, 5.5, 9.5]

media = sum(notas) / len(notas)

# Desvio padrão mede o quanto os valores se afastam da média.
# Passo a passo é calculo do quadrado da diferença de cada valor para a
# média, tirando a variância e extraindo a raiz.
soma_quadrados = 0
for nota in notas:
    soma_quadrados += (nota - media) ** 2

variancia = soma_quadrados / len(notas)      # variância
desvio_padrao = variancia ** 0.5             # raiz quadrada

print(f"Notas: {notas}\n")
print(f"Média: {media:.2f}")
print(f"Variância: {variancia:.2f}")
print(f"Desvio padrão: {desvio_padrao:.2f}")
```

## 48. Ordenação por Inserção
Implementar o algoritmo de ordenação por inserção (Insertion Sort).

```python
def insertion_sort(lista):
    numeros = lista.copy()

    for i in range(1, len(numeros)):
        atual = numeros[i]      # elemento a ser inserido no lugar certo
        j = i - 1               # último índice da parte já ordenada

        while j >= 0 and numeros[j] > atual:
            numeros[j + 1] = numeros[j]
            j -= 1

        numeros[j + 1] = atual

    return numeros

original = [29, 10, 14, 37, 13, 5]
print(f"Lista original: {original}")
print(f"Lista ordenada: {insertion_sort(original)}")
```

## 49. Soma dos Dígitos
Calcular a soma dos dígitos de um número inteiro fornecido.

```python
numero = int(input("Digite um número inteiro: "))

soma_matematica = 0
temporario = abs(numero)
while temporario > 0:
    soma_matematica += temporario % 10 # utilo o %10 para extrais o ultimo digito
    temporario //= 10 # E //10 para descartar assim até o numero zerar

print(f"Número: {numero}")
print(f"Soma dos dígitos : {soma_matematica}")
```

## 50. Cálculo de Frequência de Letras
Calcular a frequência de cada letra em uma string fornecida pelo usuário.

```python
texto = input("Digite um texto: ")

# Filtrando com .isalpha() para considerar apenas letras.
frequencia = {}
for caractere in texto.lower():
    if caractere.isalpha():
        frequencia[caractere] = frequencia.get(caractere, 0) + 1

total_letras = sum(frequencia.values())

print(f"\nTexto analisado: {texto}")
print(f"Total de letras: {total_letras}\n")

# Dessa maneira odena pelo valor do maior para o menor.
# A função lambda indica qual parte do par será usada na comparação.
ordenado = sorted(frequencia.items(), key=lambda item: item[1], reverse=True)

print("Letra | Ocorrências | Frequência relativa")
for letra, quantidade in ordenado:
    percentual = (quantidade / total_letras) * 100
    print(f"  {letra}   |      {quantidade}      |   {percentual:.1f}%")
```
